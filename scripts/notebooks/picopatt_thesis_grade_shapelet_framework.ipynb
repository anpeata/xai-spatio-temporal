```python

STAT_NAMES = [
    'mean','median','min','max','std',
    'q10','q25','q75','q90',
    'mean_abs_d1',
    'mean_abs_d2'
]

def stats11(x):
    x=np.asarray(x,dtype=float)
    x=x[~np.isnan(x)]
    if len(x)<3:
        return [np.nan]*11

    d1=np.diff(x)
    d2=np.diff(x,n=2)

    return [
        np.mean(x),
        np.median(x),
        np.min(x),
        np.max(x),
        np.std(x),
        np.quantile(x,0.10),
        np.quantile(x,0.25),
        np.quantile(x,0.75),
        np.quantile(x,0.90),
        np.mean(np.abs(d1)),
        np.mean(np.abs(d2))
    ]
```
